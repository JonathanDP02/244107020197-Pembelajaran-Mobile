```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder, StandardScaler
from sklearn.model_selection import train_test_split

df = pd.read_csv('wbc.csv')

# --- 1. Memisahkan variabel yang dapat digunakan dan tidak dapat digunakan ---
kolom_tidak_berguna = ['id', 'Unnamed: 32']
df = df.drop(columns=[col for col in kolom_tidak_berguna if col in df.columns])

# Memisahkan fitur (X) dan target (y)
X = df.drop(columns=['diagnosis'])
y = df['diagnosis']

# --- 2. Melakukan proses encoding pada kolom "diagnosis" ---
label_encoder = LabelEncoder()
y_encoded = label_encoder.fit_transform(y)

# --- 3. Melakukan proses standarisasi pada semua kolom numerik ---
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# Mengembalikan format X_scaled ke dalam DataFrame
X_scaled_df = pd.DataFrame(X_scaled, columns=X.columns)

# --- 4. Melakukan proses stratified split data (Rasio 80:20) ---
X_train, X_test, y_train, y_test = train_test_split(
    X_scaled_df,
    y_encoded,
    test_size=0.2,
    stratify=y_encoded,
    random_state=42
)

print("Data berhasil diproses!")
print(f"Jumlah Data Latih (80%): {X_train.shape[0]} baris")
print(f"Jumlah Data Uji (20%): {X_test.shape[0]} baris")
```
